```python
import pandas as pd

# Sample dataset
data = {
    'Age': [25, 30, 35, 40, None],
    'Salary': [50000, 60000, None, 80000, 90000],
    'Department': ['Sales', 'Engineering', 'HR', None, 'Marketing']
}
df = pd.DataFrame(data)
print(df)
```

```python
import numpy as np
from sklearn.impute import SimpleImputer

# Replace None in numeric columns

# Replace the Nones in Age column
numeric_imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
numeric_imputer.fit(df[['Age']])
df[['Age']] = numeric_imputer.transform(df[['Age']])

# Replace the Nones in Salary column
numeric_imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
numeric_imputer.fit(df[['Salary']])
df[['Salary']] = numeric_imputer.transform(df[['Salary']])
print(df)
```

```python
# Drop rows with None in categorical column
df.dropna(inplace=True)
print(df)
```
